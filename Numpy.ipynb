```python
a = np.array([[3,2,1],[3,1,2],[0,5,4]])

print(a.sum())
print("-----")

print(a.sum(axis = 0)) # along which direction we want to apply the aggrgate fucntion -  columns
print("-----")

print(a.sum(axis = 1)) # along which direction we want to apply the aggrgate fucntion -  rows

print("----")

print(a.max(axis=0))

print("----")

print(a.max(axis=1))

print("----")

print(a.mean(axis=1))
print("----")

```
